```python
#Q2
import numpy as np

count = 0
range_start = 1
range_end = 100

for i in range(10000):
    num = np.random.randint(range_start, range_end + 1)

    if num % 12 == 0:
        count = count + 1

print('Number of multiples of 12:', count)
```

```python
#Q3
import numpy as np

def fun_square_count():
    num_list = np.arange(1, 101)
    squred_num = [value**2 for value in num_list]
    count = 0
    for num in squred_num:
        if (num % 10 == 1):
            count += 1
    return count

result = fun_square_count()
print(f"Your result is {result}")
print(f"Expected result is {20}")
```

```python
#Q4
import numpy as np

def is_prime(n):
    if n < 2:
        return False

    for num in range(2, n):
        if n % num == 0:
            return False
    return True

numbers = np.arange(3, 21)
results = {}
for num in numbers:
    results[num] = is_prime(num)

print(f"Your results is {results}")
expected_result = {3: True, 4: False, 5: True, 6: False, 7: True, 8: False, 9: False, 10: False, 11: True, 12: False, 13: True, 14: False, 15: False, 16: False, 17: True, 18: False, 19: True, 20: False}
print(f"Expected result is: {expected_result}")
```
