Q1) Run the following programs, identify and Fix the errors.
Use exception handling approach to handle the erros.

```python
#Q1.1
def myfunction(a, b):
    try:
        answer = a / b
        return answer
    except ZeroDivisionError:
        return "You cannot divide by zero!"

print(myfunction(2, 0))
```

```python
#Q1.2
def myfunction2(c):
    try:
        answer = a / c
        return answer
    except NameError:
        return "Error: Variable 'a' is not defined."

print(myfunction2(3))
```

```python
#Q1.3
def myfunction3(b):
    mylist = [2, 3, 5, -7, 8, 0]
    idx = 0
    while idx <= 6:
        try:
            print('Iteration:', idx, mylist[idx]/b)
            idx = idx + 1
            print('Iteration completed')
        except IndexError:
            print(f"Error: Index {idx} is out of range for a list of size {len(mylist)}.")
            break

myfunction3(b=5)
```

```python
#Q1.4
def divide(a, b):
    try:
        result = a / b
        print('The result of division is', result)
        return result

    except ZeroDivisionError:
        print("Error: You cannot divide by zero.")
        return None

    except (ValueError, TypeError):
        print("Error: Please provide valid numbers. Strings cannot be divided.")
        return None

divide("10", 2)
divide(10, 0)
divide(10, "two")
```

```python
#Q1.5
def read_file(filename):
    try:
        with open(filename, 'r') as file1:
            word_string = file1.read()
        print("File reading completed successfully")
        return word_string

    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found. Please check the file path.")
        return None

read_file("Q1.txt")
```
